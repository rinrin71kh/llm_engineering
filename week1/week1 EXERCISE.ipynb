```python
# imports
import os 
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:1b'
```

```python
# set up environment
load_dotenv(override=True)
ollama_base_url = os.getenv('OLLAMA_BASE_URL')
Ollama = OpenAI(base_url=ollama_base_url , api_key = MODEL_LLAMA)
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:
         stream = Ollama.chat.completions.create(model = MODEL_LLAMA , messages = [{"role":"user" ,"content":question  }] , stream = True)
        response = ""
        for chunk in stream :
            response += chunk.choices[0].delta.content or ''
        print(response)
"""
```

```python
system_prompt = "You are a helpful technical tutor who answers questions about python code, software engineering, data science and LLMs"
user_prompt = "Please give a detailed explanation to the following question: " + question
```

```python
Messages = [{"role":"system" ,"content":system_prompt},{"role":"user" ,"content":user_prompt}]
```

```python
# Get gpt-4o-mini to answer, with streaming
```

```python
# Get Llama 3.2 to answer
# in my case i can't use chatgpt api key right now so i stick to ollama only
response = Ollama.chat.completions.create(model = MODEL_LLAMA , messages = Messages)
print(response.choices[0].message.content)
```

```python
# Get Llama 3.2 to answer with stream
# in my case i can't use chatgpt api key right now so i stick to ollama only
stream = Ollama.chat.completions.create(model = MODEL_LLAMA , messages = Messages , stream = True)
response = ""
for chunk in stream :
    response += chunk.choices[0].delta.content or ''
print(response)
```
